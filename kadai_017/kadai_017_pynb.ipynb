```python
class Human:
  def __init__(self, name, age):
    self.name=name
    self.age=age
  def check_adult(self):
    if self.age >= 20:
       print("大人")
    else:
       print("大人でない")
```

```python
human1=Human("Sato", 25)
```

```python
human2=Human("Ito", 19)
```

```python
humans={1:human1, 2:human2}
```

```python
for value in humans.values():
   value.check_adult()
```
